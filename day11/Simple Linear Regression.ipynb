```python
# Step 1: Define dataset and import matplotlib for plotting
import matplotlib.pyplot as plt
# Enable inline plotting for Jupyter Notebooks
%matplotlib inline
X = [1, 2, 3, 4, 5]
Y = [2, 4, 5, 4, 5]
n = len(X)
print("Dataset loaded successfully.")
```

```python
# Step 2: Initialize parameters and hyperparameters
m = 0.0
c = 0.0
learning_rate = 0.01
epochs = 1000
print(f"Initial guesses -> m: {m}, c: {c}")
print(f"Learning Rate: {learning_rate}, Epochs: {epochs}")
```

```python
for epoch in range(epochs):
    dm = 0.0
    dc = 0.0

    for i in range(n):
        y_pred = m * X[i] + c
        error = y_pred - Y[i]

        dm += (2 / n) * error * X[i]
        dc += (2 / n) * error

    m = m - learning_rate * dm
    c = c - learning_rate * dc

# Output
print("--- Optimization Complete ---")
print(f"Optimized Slope (m): {m:.4f}")
print(f"Optimized Intercept (c): {c:.4f}")
```

```python
# Step 4: Calculate final predictions using the optimized m and c

predictions = [m * x + c for x in X]

print("Actual Y:   ", Y)
print("Predicted Y:", [round(p, 2) for p in predictions])
```

```python
# Step 5: Calculate Mean Squared Error

mse = sum([(Y[i] - predictions[i])**2 for i in range(n)]) / n

print(f"Mean Squared Error (MSE): {mse:.4f}")
```

```python
# Step 5: Plot the actual data points and the learned regression line

import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

# Plot actual data points (Scatter plot)
plt.scatter(X, Y, color='blue', s=100, label='Actual Data Points')

# Plot the regression line
plt.plot(
    X,
    predictions,
    color='red',
    linewidth=2,
    label=f'Regression Line: y = {m:.2f}x + {c:.2f}'
)

# Add labels, title, and legend
plt.xlabel('X (Independent Variable)', fontsize=12)
plt.ylabel('Y (Dependent Variable)', fontsize=12)
plt.title('Simple Linear Regression via Gradient Descent', fontsize=14)

plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()

# Display the plot
plt.show()
```
